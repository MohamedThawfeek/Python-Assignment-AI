```python
import pickle
filename="final_model_both.sav"
load_model=pickle.load(open(filename, "rb"))

with open("sclarX.pkl", "rb") as file:
    sc = pickle.load(file)

with open("sclarY.pkl", "rb") as file:
    scy = pickle.load(file)

RD_Spend = float(input("R&D Spend: "))
Administration = float(input("Administration: "))
Marketing_Spend = float(input("Marketing Spend: "))
Florida = input("State Florida (Yes/No): ").strip().lower()
New_York = input("State New York (Yes/No): ").strip().lower()

state = 1 if Florida == "yes" else 0
state1 = 1 if New_York == "yes" else 0
final_user_value = sc.transform([[RD_Spend, Administration, Marketing_Spend, state, state1]])
result = load_model.predict(final_user_value)

final_result = scy.inverse_transform([result])

final_result = round(final_result[0][0], 2)

print(final_result)

```
