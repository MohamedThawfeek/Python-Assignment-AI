```python
import pandas as pd
dataset = pd.read_csv("50_Startups_dataset.csv")
dataset = pd.get_dummies(dataset,dtype=int, drop_first=True)
dataset = dataset.drop(columns="Unnamed: 0")

dataset
```

```python
indipendent=dataset[["R&D Spend", "Administration", "Marketing Spend", "State_Florida", "State_New York"]]
dependent=dataset[["Profit"]]
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test, y_train, y_test = train_test_split(indipendent, dependent, test_size=0.3, random_state=0)

X_test
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
X_train= sc.fit_transform(X_train)
X_test= sc.transform(X_test)
```

```python
scy=StandardScaler()
y_train= scy.fit_transform(y_train)
y_test= scy.transform(y_test)
```

```python
from sklearn.svm import SVR
regrossor= SVR(kernel="rbf", C=1000)
regrossor.fit(X_train, y_train)
```

```python
regrossor._intercept_
```

```python
regrossor.n_support_
```

```python
regrossor.support_
```

```python
y_predict=regrossor.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r2_score=r2_score(y_test,y_predict)
r2_score
```

```python
import pickle
filename="final_model_both.sav"
```

```python
pickle.dump(regrossor, open(filename, "wb"))
```

```python
with open("sclarX.pkl", "wb") as file:
    pickle.dump(sc, file)

print("Scaler saved successfully")
```

```python
with open("sclarY.pkl", "wb") as file:
    pickle.dump(scy, file)

print("Scaler saved successfully")
```
