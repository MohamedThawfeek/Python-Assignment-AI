```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from Univariate import Univariate
function = Univariate

pd.set_option('display.max_columns', None)
dataset = pd.read_csv("kidney_disease.csv")

# Get the dataset some time numerical values comming objectTypes so we need to convert the colunms to numerical datatypes
Int = ['pcv', 'wc', 'rc', 'age', 'bp', 'sg', 'al', 'su', 'bgr', 'bu', 'sc', 'sod', 'pot', 'hemo']  # example columns
for col in Int:
    dataset[col] = pd.to_numeric(dataset[col], errors='coerce')


quan,qual= function.quanQual(dataset)

# Numarical Values convert
num = SimpleImputer(missing_values=np.nan, strategy='mean')
num.fit(dataset[quan])
numericalValues=num.transform(dataset[quan])

# Categorical Values convert
cat = SimpleImputer(missing_values=np.nan, strategy='most_frequent')
cat.fit(dataset[qual])
categoricalValues=cat.transform(dataset[qual])


# Convert NumPy arrays to DataFrames before concatenating
numerical_df = pd.DataFrame(numericalValues, columns=quan)
categorical_df = pd.DataFrame(categoricalValues, columns=qual)

# Now concatenate the DataFrames
dataset = pd.concat([numerical_df, categorical_df], axis=1)

# Fix pandas display settings for wide DataFrames
pd.set_option('display.expand_frame_repr', False)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 2000)

# Display the DataFrame

dataset.to_csv("pre_process_dataset.csv", index=False)

```
