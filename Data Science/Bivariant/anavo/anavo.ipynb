```python


def two_way_anova(dataset, factor1, factor2, value_col):
    import statsmodels.api as sm
    from statsmodels.formula.api import ols

    # formula value_col is calculation factor1 and factor2 match the table column
    formula = f"{value_col} ~ C({factor1}) * C({factor2})"

    #Ordinary Least Squares regression model build inside given calculation formula and dataset fit method take best 
    #regression line give model output
    model = ols(formula, data=dataset).fit()

    # ANOVA table inside given model type have 3 method. 1,2,3 and type 1 is first variable influence, 
    # type 2 is each variable effect and interaction consider given type two is best 
    # type 3 is used for most compled model and advance usage
    anova_table = sm.stats.anova_lm(model, typ=2)

    return anova_table

result = two_way_anova(dataset, "gender", "degree_t", "salary")

print(result)
```
