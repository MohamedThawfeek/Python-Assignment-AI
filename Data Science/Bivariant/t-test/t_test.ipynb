```python
import pandas as pd
import seaborn as sns
from Univariate import Univariate
from scipy.stats import ttest_ind
# Load dataset
dataset = pd.read_csv("PreProcess.csv")




# ttest_ind(male, female)
def t_test_cal(dataset, filter_col, val1, val2, target_col, target_col2):
    group1 = dataset[dataset[filter_col] == val1][target_col].dropna()
    group2 = dataset[dataset[filter_col] == val2][target_col2].dropna()

    t_stat, p_value = ttest_ind(group1, group2)

    return {
        "column":filter_col,
        "filter": val1,
        "filter2": val2,
        "target_col": target_col,
        "target_col2": target_col2,
        "t_stat": t_stat,
        "p_value": p_value
    }


t_test_cal(dataset, "gender", "M", "F", "salary", "salary")
```
