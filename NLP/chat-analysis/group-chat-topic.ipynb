```python
from sklearn.feature_extraction.text import TfidfVectorizer, ENGLISH_STOP_WORDS
import pandas as pd
from sklearn.decomposition import NMF
```

```python
dataset = pd.read_csv("group_final_data.csv")
```

```python
custom_words = {
    "hai", "hello", "hi", "ok", "okay",
    "yes", "yeah", "thanks", "please",
    "media", "omitted", "deleted", "message", "bro", "ena", "panalama", "bri"
}
extended_stop_words = list(ENGLISH_STOP_WORDS.union(custom_words))
tfidf = TfidfVectorizer(max_df=0.95, min_df=2, stop_words=extended_stop_words)
dtm = tfidf.fit_transform(dataset["Chat"])
```

```python
nmf_model = NMF(n_components=10, random_state=42)
nmf_model.fit(dtm)

for index, topic in enumerate(nmf_model.components_):
    results = ([tfidf.get_feature_names_out()[i] for i in topic.argsort()[-10:]])
    # print(results)

topic_results=nmf_model.transform(dtm)
dataset["Topic"]= topic_results.argmax(axis=1)

```

```python
# Want to save that CSV file use this command 
dataset.to_csv("group_final_data.csv", index=False)
```
