```python
from ollama import Client, Options
from getpass import getpass
import os

api_key = getpass("Enter your new Gemini API key: ")
host="https://ollama.com"
client = Client(host, headers={"Authorization": f"Bearer {api_key}"})
```

```python
def get_completion_from_messages(messages, model="gpt-oss:120b", temperature=0, max_tokens=200):
    response= client.chat(model,messages=messages, think="low", options=Options(temperature=temperature, num_predict=max_tokens))

    content = response.message.content
    thinking = response.message.thinking
    token_dict = {
        "prompt_tokens": response.prompt_eval_count or 0,
        "completion_tokens": response.eval_count or 0,
        "total_tokens": ((response.prompt_eval_count or 0)+ (response.eval_count or 0))

    }
    return content, thinking, token_dict, response
```

```python
delimiter="####"
system_message=f""" 
you will be provided with customer service queries.
the customer service query will be delimited with
{delimiter} characters.
classify each query into a primary category and a secondary category
provide your output in json format with the
keys: primary and secondary.

primary categories: billing, technica support, account management or generate inquiry.

billing secondary categories:
unsubscribe or upgrade
add a payment method
explanation for charge
dispute a charge

techinal support secondary categories:
general troubleshooting
device compatibility
software updates

account management secondary categories:
password reset
update personal information
close account
accoutn security

general inquiry secondary categories:
product information
pricing
feedback
speak to a human
"""

user_message=f""" i need motrola edge60 smarphone information """
```

```python
messages= [
    {"role": "system", "content": system_message},
    {"role": "user", "content": f"{delimiter}{user_message}{delimiter}"}
]

content, thinking, token_dict, response = get_completion_from_messages(messages, temperature=0)
print(content)
```
