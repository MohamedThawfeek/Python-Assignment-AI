```python
from ollama import Client, Options
from getpass import getpass
import os

api_key = getpass("Enter your new Gemini API key: ")
host="https://ollama.com"
client = Client(host, headers={"Authorization": f"Bearer {api_key}"})
```

```python
def get_completion(prompt, model="gpt-oss:120b"):
    messages=[{"role": "user", "content":prompt}]
    response= client.chat(model,messages=messages, options=Options(temperature=0))
    reply = response.message.content
    return reply
```

```python
response = get_completion("what is the capital of france ?")
print("\n" + response + "\n")
```

```python
response = get_completion("take the letter in lollipop and reverse them")
print("\n" + response + "\n")
```

```python
response = get_completion("take the letter in l-o-l-l-i-p-o-p and reverse them")
print("\n" + response + "\n")
```

```python
def get_completion_from_messages(messages, model="gpt-oss:120b", temperature=0, max_tokens=500):
    response= client.chat(model,messages=messages, options=Options(temperature=temperature, num_predict=max_tokens))
    reply = response.message.content
    return reply
```

```python
messages= [
    {"role": "system", "content": """ you are an assistant who responds in the style Dr Seuss. """},
    {"role": "user", "content": """ write me a very short poem about a happy carrot """}
]

response = get_completion_from_messages(messages, temperature=1)
print("\n" + response + "\n")
```

```python
def get_completion_from_messages(messages, model="gpt-oss:120b", temperature=0, max_tokens=200):
    response= client.chat(model,messages=messages, think="low", options=Options(temperature=temperature, num_predict=max_tokens))

    content = response.message.content
    thinking = response.message.thinking
    token_dict = {
        "prompt_tokens": response.prompt_eval_count or 0,
        "completion_tokens": response.eval_count or 0,
        "total_tokens": ((response.prompt_eval_count or 0)+ (response.eval_count or 0))

    }
    return content, thinking, token_dict, response
```

```python
messages= [
    {"role": "system", "content": """ you are an assistant who responds in the style Dr Seuss. """},
    {"role": "user", "content": """ write me a very short poem about a happy carrot """}
]

content, thinking, token_dict, response = get_completion_from_messages(messages, temperature=0)
print(content)
```
