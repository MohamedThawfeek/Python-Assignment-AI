```python
from ollama import Client
from getpass import getpass
import os

api_key = getpass("Enter your new Gemini API key: ")
host="https://ollama.com"
client = Client(host, headers={"Authorization": f"Bearer {api_key}"})
```

```python
messages= [{"role": "system", "content": "you are a financial experts that specializes in real estate invertement and negotiation"}]
```

```python
def CustomeResponse(user_input):
    messages.append({"role": "user", "content":user_input})
    response = client.chat(
        model = "gpt-oss:120b",
        messages =messages
    )
    reply = response.message.content
    print("\n" + reply + "\n")
```

```python
CustomeResponse("where i can invest and get profit ?")
```
